```python
import shutil
import gzip
import os

src_path = "/data/jx4237data/PKG2TKG/viz_notebook/author_embeddings.pkl.gz"
dst_dir = "/data/jx4237data/PKG2TKG/cfde_demo/data"
dst_gz_path = os.path.join(dst_dir, "author_embeddings.pkl.gz")
dst_pkl_path = os.path.join(dst_dir, "author_embeddings.pkl")

# Ensure the destination directory exists
os.makedirs(dst_dir, exist_ok=True)

# Copy the .gz file
shutil.copy(src_path, dst_gz_path)

# Unzip the .gz file to .pkl
with gzip.open(dst_gz_path, "rb") as f_in, open(dst_pkl_path, "wb") as f_out:
    shutil.copyfileobj(f_in, f_out)

print(f"Copied {src_path} to {dst_gz_path} and extracted to {dst_pkl_path}")
```

```python
import os
import gzip

base_dir = "/data/jx4237data/PKG2TKG/deliverable_data"
extensions = [".csv", ".tsv", ".csv.gz", ".tsv.gz"]

for root, dirs, files in os.walk(base_dir):
    for fname in files:
        for ext in extensions:
            if fname.endswith(ext):
                fpath = os.path.join(root, fname)
                # 打开文件并读取表头
                if ext.endswith(".gz"):
                    with gzip.open(fpath, "rt", encoding="utf-8", errors="ignore") as f:
                        try:
                            header = next(f).strip()
                        except StopIteration:
                            header = ""
                        n_lines = 1 + sum(1 for _ in f) if header else 0
                else:
                    with open(fpath, "rt", encoding="utf-8", errors="ignore") as f:
                        try:
                            header = next(f).strip()
                        except StopIteration:
                            header = ""
                        n_lines = 1 + sum(1 for _ in f) if header else 0
                print(f"{fpath}: {n_lines} lines")
                print(f"Header: {header}")
```
